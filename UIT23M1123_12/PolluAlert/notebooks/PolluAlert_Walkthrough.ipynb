### 5. Model Evaluation Metrics Inspection

```python
eval_path = os.path.join('..', 'models', 'evaluation_results.json')
if os.path.exists(eval_path):
    with open(eval_path, 'r') as f:
        eval_data = json.load(f)
    for task, models in eval_data.items():
        print(f"\n=== Task: {task.upper()} ===")
        for model_name, metrics in models.items():
            print(f"  {model_name:22s} -> Acc: {metrics['accuracy']:.4f} | F1: {metrics['f1_macro']:.4f}")
```

### 6. End-to-End Live Complaint Prediction Demonstration

```python
test_complaints = [
    "Illegal chemical waste dumping into Bellandur Lake near Whitefield causing fish death and strong stench.",
    "Massive traffic noise from continuous honking and modified silencers on MG Road crossing 90 decibels.",
    "CRITICAL: Ammonia gas leak from fertilizer plant near residential zone, children hospitalized!",
    "Municipal garbage bin overflowing on 5th main street for the past 4 days without clearance.",
    "Enjoying a beautiful sunny morning in Cubbon park with pleasant weather."
]

for idx, text in enumerate(test_complaints, 1):
    result = predict_complaint(text)
    print(f"\n------------------- Sample {idx} -------------------")
    print(f"Input Text:   {text}")
    print(f"Pollution:    {result['pollution_type']} (Confidence: {result['pollution_confidence']:.2f})")
    print(f"Severity:     {result['severity']} (Score: {result['severity_score']})")
    print(f"Emergency:    {'🚨 YES' if result['is_emergency'] else '✅ NO'} ({result['emergency_label']})")
    print(f"Locations:    {result['entities']['locations']}")
    print(f"Action Alert: {result['recommended_action']}")
```
